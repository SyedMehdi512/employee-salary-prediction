# Drop Irrelevant Features

# Convert date and extract features

```python
df['bdate'] = pd.to_datetime(df['bdate'])

current_year = pd.Timestamp.now().year

df['age'] = current_year - df['bdate'].dt.year
```

```python
df = df.drop(columns=['bdate'])
```

```python
# Convert obje to num
df['prevexp'] = pd.to_numeric(df['prevexp'], errors='coerce')
```

## Handle missing values

```python
from sklearn.impute import SimpleImputer
num_cols = ['educ', 'salary', 'salbegin', 'jobtime', 'prevexp', 'age']

num_imputer = SimpleImputer(strategy='median')

df[num_cols] = num_imputer.fit_transform(df[num_cols])
```

```python
cat_cols = ['gender', 'jobcat', 'minority']

cat_imputer = SimpleImputer(strategy='most_frequent')

df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])
```

```python
df.isnull().sum()
```

```python
import numpy as np
```

## Handle outliers (IQR capping)

```python
for col in ['salary', 'salbegin', 'prevexp', 'jobtime']: #10k, 10k, 10k, 50k-120k, 170k
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR #10k
    upper = Q3 + 1.5 * IQR #170k
    
    df[col] = np.clip(df[col], lower, upper)
```

# Feature Engineering

### Salary growth feature

```python
df['salary_growth'] = df['salary'] - df['salbegin']
```

### Experience ratio feature

```python
df['experience_ratio'] = df['prevexp'] / (df['jobtime'] + 1)
```

### Education category (binning)

```python
df['educ_level'] = pd.cut(df['educ'],
                          bins=[0,12,16,25],
                          labels=['School','Undergrad','Graduate'])
```

```python
df.head(3)
```

### Age group feature

```python
df.age.max()
```

```python
df.age.min()
```

```python
df['age_group'] = pd.cut(df['age'],
                         bins=[55,75,85,97],
                         labels=['Young','Mid','Senior'])
```
